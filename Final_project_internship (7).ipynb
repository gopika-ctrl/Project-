CALCULATE TOTAL MARKS

```python
subjects=['Maths','Science','English','Computer','Social']
df['total']=df[subjects].sum(axis=1)
df[['Student_ID','total']].tail()
```

CALCULATE AVERAGE MARKS

```python
df['average']=df[subjects].mean(axis=1)
df[['Student_ID','average']].head(8)
```

TOTAL DATASET COLUMNS

```python
df.columns
```

ASSIGNING GRADES

```python
def grade(avg):
  if avg >=90:
   return'A+'
  elif avg >=80:
   return'A'
  elif avg >=70:
   return'B'
  elif avg >=60:
   return'C'
  elif avg >=50:
   return'D'
  else:
   return'F'
df['Grade'] = df['average'].apply(grade)
df[['Student_ID','average','Grade']].tail(7)
```

BAR CHART-AVERAGE MARKS BY SUBJECT

```python
import matplotlib.pyplot as plt
subject_avg = df[subjects].mean()

subject_avg.plot(kind='bar', color=['red','blue','orange','yellow','pink'])

plt.title('Average Marks by Subject')
plt.xlabel('Subjects')
plt.ylabel('Average Marks')
plt.show()
```

PIE CHART-PASS VS FAIL

```python
import matplotlib.pyplot as plt

df['Status'] = df['Grade'].apply(lambda x: 'Pass'if x != 'F'else'Fail')

df['Status'].value_counts().plot(
 kind='pie',
 autopct='%1.1f%%'
)

plt.title('Pass vs Fail')
plt.ylabel('')

plt.show()
```

TOTAL MARKS DISTRIBUTION

```python
import matplotlib.pyplot as plt
df['total'] = df[['Maths','Science','English','Computer','Social']].sum(axis=1)
plt.hist(df['total'], bins=10)

plt.title("Total Marks Distribution")
plt.xlabel("Total Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df['English'], df['Computer'])

plt.title("English vs Computer")
plt.xlabel("English Marks")
plt.ylabel("Computer Marks")

plt.show()
```
