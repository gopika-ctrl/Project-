calculate total marks

```python
subjects=['Maths','Science','English','Computer','Social']
df['total']=df[subjects].sum(axis=1)
df[['Student_ID','total']].head(7)
```

calculate average marks

```python
df['average']=df[subjects].mean(axis=1)
df[['Student_ID','average']].head(7)
```

assigning grades

```python
def grade(avg):
  if avg>=80:
    return'A+'
  elif avg>=70:
    return'A'
  elif avg>=60:
    return'B'
  elif avg>=55:
    return'C'
  elif avg>=50:
    return'D'
  else:
    return'F'
df['grade']=df['average'].apply(grade)
df[['Student_ID','average','grade']].head(7)
```
